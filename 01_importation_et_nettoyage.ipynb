## 3. Nettoyage des données
### 3.1. Nettoyage des colonnes textuelles

L'objectif de cette étape est d'éliminer le "bruit" invisible dans les colonnes de type `object` :
- Suppression des espaces superflus au début et à la fin (`.str.strip()`).
- Choix de conserver la casse d'origine (pas de `.str.lower()`), les libellés étant déjà propres et professionnels pour un rapport B2B.

```python
# Sélectionne toutes les colonnes textuelles
colonnes_texte = Sales.select_dtypes(include=["object"]).columns
# Applique le strip() sur chacune d'elles
for cols in colonnes_texte:
    Sales[cols] = Sales[cols].str.strip()
print("Espace avant et après supprimés avec succès")
```

### 3.2. Conversion et enrichissement de la date de commande (`OrderDate`)

- **Conversion** : Passage du format texte au format `datetime` de Pandas pour permettre les tris et analyses chronologiques.
- **Enrichissement (Feature Engineering)** : Extraction de nouvelles colonnes clés pour le reporting (Année, Mois, Trimestre, etc.).

```python
# Conversion de la colonne OrderDate au format datetime
Sales["OrderDate"] = pd.to_datetime(Sales["OrderDate"])
print("Date convertie au format Date")
```

```python
print("Extraction de l'année...")
Sales["OrderYear"] =  Sales["OrderDate"].dt.year

print("Extraction du mois...")
Sales["OrderMonth"] = Sales["OrderDate"].dt.month

print("Extraction du trimestre...")
Sales["OrderQuarter"] = Sales["OrderDate"].dt.quarter

print("Enrichissement temporel terminé avec succès !")
```

```python
# Vérification rapide des dates
print("Valeurs manquantes dans OrderDate :", Sales["OrderDate"].isnull().sum())
print("Période couverte : du", Sales["OrderDate"].min(), "au", Sales["OrderDate"].max())
```

### 3.3. Vérification des colonnes numériques
- Les types (`int`) et les plages de valeurs ont été audités lors de l'EDA. 
- Aucune anomalie (valeurs négatives ou aberrantes) détectée, aucune transformation nécessaire sur cette partie.

```python
# vérification des limites sur une colonne de quantité
# On cherche les lignes qui violent la règle (quantité <= 0 ou quantité > 25)
anomalies_quantity = Sales[(Sales["Quantity"] < 0) | (Sales["Quantity"] > 25)]
print(f"Nombre d'anomalies détectées sur les quantités : {len(anomalies_quantity)}")

# Si besoin d'afficher ces lignes pour les inspecter :
if len(anomalies_quantity) > 0:
    display(anomalies_quantity)
else:
    print("Toutes les quantités respectent les règles métiers.")
```

```python
# vérification des limites sur une colonne de quantité
# On cherche les lignes qui violent la règle (quantité <= 100 ou quantité > 10000)
anomalies_UnitPrice = Sales[(Sales["UnitPrice"] <= 20) |  (Sales["UnitPrice"] > 10000)]
print(f"Nombre d'anomalies détectées sur les UnitPrice : {len(anomalies_UnitPrice)}")

if len(anomalies_UnitPrice) > 0:
    display(anomalies_UnitPrice)
else: print("Tous les UnitPrice respectent les règles métiers.")
```
