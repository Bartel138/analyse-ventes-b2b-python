```python
Sales = pd.read_csv("Sales _bronze_silver.csv")
print("Fichier importé avec sucès")
```

## 2.3. Analyse de la structure et des types de données (info)

```python
print("Resume statistique : \n")
display(Sales.describe())
```

## 2.5. Détection précise des valeurs manquantes (isnull().sum)

```python
print("Analyse détaillée des valeurs manquantes :\n")

valeurs_manquantes = pd.DataFrame({
    "Valeurs manquantes": Sales.isnull().sum(),
    "Pourcentage %": (Sales.isnull().sum() / len(Sales)) * 100
})

display(valeurs_manquantes)
```

## 2.6. Recherche des lignes dupliquées (duplicated().sum)

```python
print("Analyse des doublons :\n")

nb_doublons = Sales.duplicated().sum()
pourcentage_doublons = (nb_doublons / len(Sales)) * 100

print(f"Nombre de lignes dupliquées : {nb_doublons}")
print(f"Pourcentage de doublons : {pourcentage_doublons:.2f}%")
```
